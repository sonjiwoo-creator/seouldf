```python
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 2020~2024 필터링
전체_추이 = (
    sales[sales['연도'].between(2020, 2024)]
    .groupby('연도')['당월_매출_금액']
    .sum()
    / 1e12
)

fig, ax = plt.subplots(figsize=(12, 6))

ax.plot(전체_추이.index, 전체_추이.values,
        marker='o', linewidth=2.5,
        color='#4C72B0', markersize=8)

ax.axvspan(2020, 2021, alpha=0.1, color='gray', label='코로나')

# 시작점 끝점 수치 표시
ax.annotate(f'{전체_추이[2020]:.0f}조',
            xy=(2020, 전체_추이[2020]),
            xytext=(2020.3, 전체_추이[2020]+0.5),
            ha='center', fontsize=11, color='#4C72B0')
ax.annotate(f'{전체_추이[2024]:.0f}조',
            xy=(2024, 전체_추이[2024]),
            xytext=(2024, 전체_추이[2024]+0.5),
            ha='center', fontsize=11, color='#4C72B0')

ax.set_title('서울 전체 상권 연도별 총매출 추이', fontsize=15)
ax.set_xlabel('연도')
ax.set_ylabel('총매출 (조 원)')
ax.set_xticks([2020, 2021, 2022, 2023, 2024])
ax.set_xticklabels(['2020', '2021', '2022', '2023', '2024'])
ax.legend()
ax.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
